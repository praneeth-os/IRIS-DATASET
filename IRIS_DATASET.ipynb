```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/iris.csv')
display("Dataset has been read")
```

```python
print("DATASET INFORMATION")
df.info()
```

```python
print("First 5 Rows:")
display(df.head())
```

```python
print("Last 5 Rows:")
display(df.tail())
```

```python
print("SHAPE OF DATASET")
print(df.shape)
```

```python
print("NULL VALUES")
print(df.isnull().sum())
```

```python
print("COLUMN NAMES")
print(df.columns)
```

```python
iris_numeric = df.select_dtypes(include=['int64', 'float64'])

print("NUMERICAL COLUMNS")
display(iris_numeric)
```

```python
corr_matrix = iris_numeric.corr()

print("CORRELATION MATRIX")
print(corr_matrix)
```

```python
print("CORRELATION HEATMAP")

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
print("DESCRIPTIVE STATISTICS")
print(iris_numeric.describe())
```

```python
print("VARIANCE")
print(iris_numeric.var())
```

```python
print("STANDARD DEVIATION")
print(iris_numeric.std())
```

```python

Q1 = iris_numeric.quantile(0.25)
Q3 = iris_numeric.quantile(0.75)

print("FIRST QUARTILE (Q1)")
print(Q1)

print("THIRD QUARTILE (Q3)")
print(Q3)

coefficient_of_quartile_deviation = (Q3 - Q1) / (Q3 + Q1)

print("COEFFICIENT OF QUARTILE DEVIATION")
print(coefficient_of_quartile_deviation)
```

```python
mean = iris_numeric.mean()

print("MEAN")
print(mean)
```

```python
mean_deviation = abs(iris_numeric - mean).mean()
print("MEAN DEVIATION")
print(mean_deviation)

coefficient_of_mean_deviation = mean_deviation / mean

print("COEFFICIENT OF MEAN DEVIATION")
print(coefficient_of_mean_deviation)
```

```python
cv = (iris_numeric.std() / iris_numeric.mean()) * 100

print("COEFFICIENT OF VARIATION")
print(cv)
```

```python
max_value = iris_numeric.max()
min_value = iris_numeric.min()

range_value = max_value - min_value

print("MAXIMUM VALUE")
print(max_value)

print("MINIMUM VALUE")
print(min_value)
```

```python
print("RANGE")
print(range_value)

coefficient_of_range = (max_value - min_value) / (max_value + min_value)

print("COEFFICIENT OF RANGE")
print(coefficient_of_range)
```
